```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

print("✅ Basic libraries ready!")
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print("Total pages:", len(documents))
print("Total chunks:", len(chunks))
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model_1 = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5"
)

print("✅ Embedding Model 1 loaded!")
```

```python
from langchain_community.vectorstores import Chroma

vector_store = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model_1,
    collection_name="research_papers_bge"
)

print("✅ Vector database created successfully!")
print("📚 Chunks indexed:", len(chunks))
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model_2 = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("✅ Embedding Model 2 loaded successfully!")
```

```python
from langchain_community.vectorstores import Chroma

vector_store_2 = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model_2,
    collection_name="research_papers_minilm"
)

print("✅ Model 2 vector database created!")
print("📚 Chunks indexed:", len(chunks))
```

```python
query = "What is the Transformer architecture?"

# Model 1 - BGE-small
results_1 = vector_store.similarity_search(query, k=3)

# Model 2 - MiniLM
results_2 = vector_store_2.similarity_search(query, k=3)

print("🔎 QUERY:", query)

print("\n========== MODEL 1: BGE-SMALL ==========")

for i, doc in enumerate(results_1, 1):
    print(f"\n--- Result {i} ---")
    print("Paper:", doc.metadata.get("paper_title"))
    print("Page:", doc.metadata.get("page_label"))
    print("Content:", doc.page_content[:250].replace("\n", " "))

print("\n========== MODEL 2: MINILM ==========")

for i, doc in enumerate(results_2, 1):
    print(f"\n--- Result {i} ---")
    print("Paper:", doc.metadata.get("paper_title"))
    print("Page:", doc.metadata.get("page_label"))
    print("Content:", doc.page_content[:250].replace("\n", " "))
```

```python
test_queries = [
    "What is the Transformer architecture?",
    "What is retrieval augmented generation?",
    "How does GPT-3 perform few-shot learning?",
    "What is the main contribution of BERT?",
    "How does RETRO use retrieval?"
]

for query in test_queries:
    print("\n" + "=" * 70)
    print("QUERY:", query)
    
    results_1 = vector_store.similarity_search(query, k=3)
    results_2 = vector_store_2.similarity_search(query, k=3)
    
    print("\nBGE-SMALL:")
    for i, doc in enumerate(results_1, 1):
        print(
            f"{i}. {doc.metadata.get('paper_title')} "
            f"| Page {doc.metadata.get('page_label')}"
        )
    
    print("\nMINILM:")
    for i, doc in enumerate(results_2, 1):
        print(
            f"{i}. {doc.metadata.get('paper_title')} "
            f"| Page {doc.metadata.get('page_label')}"
        )
```

```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)

print("✅ Reranker loaded successfully!")
```

```python
query = "What is the Transformer architecture?"

# Step 1: Retrieve more candidates using hybrid search
candidate_docs = hybrid_retriever.invoke(query)

# Take top 10 candidates
candidate_docs = candidate_docs[:10]

# Step 2: Create query-document pairs for reranking
pairs = [
    [query, doc.page_content]
    for doc in candidate_docs
]

# Step 3: Get relevance scores
scores = reranker.predict(pairs)

# Step 4: Combine documents with scores
ranked_results = sorted(
    zip(candidate_docs, scores),
    key=lambda x: x[1],
    reverse=True
)

# Step 5: Select best 3
final_results = ranked_results[:3]

print("🔎 Query:", query)
print("\n🎯 RERANKED TOP 3 SOURCES:\n")

for i, (doc, score) in enumerate(final_results, start=1):
    print(f"--- Result {i} ---")
    print("Score:", round(float(score), 4))
    print("Paper:", doc.metadata.get("paper_title"))
    print("Page:", doc.metadata.get("page_label"))
    print("Content:", doc.page_content[:300].replace("\n", " "))
    print()
```

```python
test_rag_queries = [
    "What is the Transformer architecture?",
    "What is retrieval augmented generation?",
    "How does GPT-3 perform few-shot learning?",
    "What is the main contribution of BERT?",
    "How does RETRO use retrieval?"
]

for query in test_rag_queries:

    # Hybrid retrieval
    candidate_docs = hybrid_retriever.invoke(query)
    candidate_docs = candidate_docs[:10]

    # Reranking
    pairs = [
        [query, doc.page_content]
        for doc in candidate_docs
    ]

    scores = reranker.predict(pairs)

    ranked_results = sorted(
        zip(candidate_docs, scores),
        key=lambda x: x[1],
        reverse=True
    )

    final_results_test = ranked_results[:3]

    # Build context
    context = "\n\n".join(
        [
            f"""
SOURCE {i}
Paper: {doc.metadata.get('paper_title')}
Page: {doc.metadata.get('page_label')}

Content:
{doc.page_content}
"""
            for i, (doc, score) in enumerate(final_results_test, 1)
        ]
    )

    # RAG prompt
    prompt = f"""
You are a research paper assistant.

Answer the question using ONLY the provided research paper context.

Do not use outside knowledge.
Do not invent information.

If the answer cannot be found in the context, say:
"I could not find the answer in the provided research papers."

Question:
{query}

Research Paper Context:
{context}

Give a clear and concise answer.
"""

    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "gemma3",
            "prompt": prompt,
            "stream": False
        }
    )

    answer = response.json()["response"]

    print("\n" + "=" * 80)
    print("QUESTION:", query)
    print("\nANSWER:")
    print(answer)

    print("\nTOP 3 SOURCES:")
    for i, (doc, score) in enumerate(final_results_test, 1):
        print(
            f"{i}. {doc.metadata.get('paper_title')} "
            f"| Page {doc.metadata.get('page_label')} "
            f"| Score {float(score):.4f}"
        )
```
